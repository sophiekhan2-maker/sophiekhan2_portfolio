```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://www.worldometers.info/world-population/population-by-country/'
```

```python
page = requests.get(url)
soup = BeautifulSoup(page.text, 'html')
```

```python
table = soup.find('table', 'datatable w-full border border-zinc-200')
```

```python
world_column = table.find_all('th')
```

```python
column_titles = [column.text for column in world_column]
print(column_titles)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = column_titles)
```

```python
df
```

```python
world_row = table.find_all('tr')
```

```python
for rows in world_row[1:]:
    row_data = rows.find_all('td')
    individual_row = [data.text for data in row_data]
    
    length = len(df)
    df.loc[length] = individual_row 
df
```

```python
df.to_csv(r'C:\Users\sophi\Documents\Python lessons\csv_folder\world_population', index = False)
```
